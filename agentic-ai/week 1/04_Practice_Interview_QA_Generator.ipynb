```python
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate, SystemMessagePromptTemplate, HumanMessagePromptTemplate
from langchain_core.output_parsers import StrOutputParser
from dotenv import load_dotenv
import os
```

```python
load_dotenv()
api_key = os.getenv("OPENAI_API_KEY")
open_ai_model = os.getenv("OPEN_AI_MODEL")
```

```python
prompt_template = ChatPromptTemplate.from_messages([
    SystemMessagePromptTemplate.from_template("""You are an expert technical interviewer and HR panelist with 10 years 
of experience designing interview questions for hiring processes across industries. 
Your questions are clear, relevant, and appropriately challenging for the stated 
difficulty level. Always write in a professional, neutral tone."""),
    HumanMessagePromptTemplate.from_template("""
                                             Generate a top 10 interview questions for preparation
                                             Job Role: {job_role}
                                             Skill Area to Focus On: {skill_area}
                                             Difficult Leven: {difficulty}
                                             Requirments:
                                             - Generate 5 questions total
                                            - Order them from easiest to hardest
                                            - For each question, add a one-line note on what a good answer should cover
                                            - Avoid yes/no questions — favor questions that reveal reasoning and experience
                                            - Use clear, simple language
                                             """)]
)
```

```python
llm = ChatOpenAI(api_key=api_key,
           model=open_ai_model,
           temperature=0.5)
```

```python
output_parser = StrOutputParser()

chain = prompt_template | llm | output_parser

role_input = {
    "job_role"   : "Gen AI developer",
    "skill_area" : "AI, ML, Agent",
    "difficulty" : "Intermediate"
}

response = chain.invoke(role_input)
print("="*50)
print()
print(response)
print("="*50)
```
